```python
class Node:
  def __init__(self, value):
    self.value = value
    self.next = None

```

```python
class Circular_ll:
  def __init__(self):
    self.head = None
    self.length = 0

  def push (self, value):#It will add to the begining
    new_node = Node(value)
    if self.head:#The linked list is not empty
      temp = self.head
      while temp.next != self.head:
        temp = temp.next
      else:
        new_node.next = self.head
        self.head = new_node
        temp.next = new_node
        self.length += 1
    else:#The linked list is empty
      self.head = new_node
      new_node.next = new_node
      self.length += 1


  def __repr__(self):
    _str = ''
    temp = self.head
    while temp.next != self.head:
      _str += str(temp.value) + "->"
      temp = temp.next
    else:
      _str += str(temp.value) + "-> [LOOP]"
    return _str

  def len(self):
    return self.length

  def is_empty(self):
    return self.length == 0

  def peek(self):#It will return the first Node
    if self.head:
      return self.head.value
    else:
      return None


  def pop(self, index):
    if self.head:

      if index == -1:#It will delete from the last
        temp = self.head
        while temp.next.next != self.head:
          temp = temp.next
        else:
          value = temp.next.value
          if self.length == 1:
            self.head = None
          else:
            temp.next = self.head
          self.length -= 1
          return value

      elif index <= 0:#IT will delete from the begining
        temp = self.head
        while temp.next != self.head:
          temp = temp.next
        else:
          value = self.head.value
          temp.next = self.head.next
          self.head = self.head.next
          self.length -= 1
          return value

      else:#It will delete from the index which is given
        temp = self.head
        if index > self.length -1:#It will delete from the last
          index = self.length -1
        index -= 1
        while index:
          temp = temp.next
          index -= 1
        else:
          value = temp.next.value
          temp.next = temp.next.next
          self.length -= 1
          return value

    else:
      return None


  def append(self, value):#It will add to the begining
    self.push(value)


  def setter(self, lst):
    l = len(lst)
    for i in lst[::-1]:
      self.push(i)


```

```python
ll = Circular_ll()
ll.push(1)
ll.append(2)
ll.push(3)
ll.push(4)
print(ll)
ll.pop(-1)
print(ll)
ll.len()
ll.setter([7,6,5])
print(ll)
ll.len()
```
