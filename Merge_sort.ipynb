```python
#Merge sort algorithm
from random import randint as rnd
amount = 10
_list = [rnd(1, amount) for i in range(amount)]
print(_list)
```

```python
def merge_lists(lst1, lst2):
  lst = []
  ln1 = len(lst1)
  ln2 = len(lst2)
  i = j = 0
  while i < len(lst1) and j < len(lst2):
    if lst1[i] < lst2[j]:
      lst.append(lst1[i])
      i += 1
    else:
      lst.append(lst2[j])
      j += 1
  lst += lst1[i:]
  lst += lst2[j:]
  return lst
```

```python
def merge_sort(_lst):
  if len(_lst) == 1:
    return _lst
  else:
    mid = len(_lst) // 2
    left = _lst[mid:]
    right = _lst[:mid]
  left = merge_sort(left)
  right = merge_sort(right)
  return merge_lists(left, right)
```
